## Research Question

How does distance to the nearest UC campus, said UC campus’s prestige, total bedrooms, and ocean proximity using the 1990 California housing dataset predict median house value, after controlling for median household income, housing age, and population density? 
To answer this, we plan to find the Euclidean distance (in miles) between each census block and the nearest UC campus using geographic coordinates. UC Merced is excluded due to its post-1990 founding. We will then find the archived rank of these Universities from the 1990 U.S. news College ranking and create a multi-variable regression model to estimate the relationship between these four predictors and median house value.


## Background and Prior Work

Housing prices depend on many different factors such as location, income, and housing size. Universities are an important local feature because they attract students, jobs, and other economic activity. In this project, we want to study how housing prices are affected by proximity to a University of California (UC) campus and the prestige of that campus. We also want to examine how well housing prices can be predicted using these factors along with other features such as the number of rooms and proximity to the coastline. 

After researching, we found that UC Riverside looked into a similar question like ours. A group of computer scientists discovered that areas with universities generally have higher home prices and rents than areas without universities.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) However, this study also found that prices near universities tend to rise and fall more quickly. This shows that while universities can influence housing prices, they tend not to be very stable. Our project builds on this work by studying how proximity to UC campuses and campus prestige relate to housing prices in California.

Previous studies have also looked at how location affects housing prices. Mattera and Franses used methods that group areas based on where they are and how housing prices in those areas move over time. They discovered that models that include location information do a better job at predicting house prices than models that choose to ignore it.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) This also shows how housing prices are influenced not just by house features, but also by where the house is and what is around it. Because of this, we are using distance to UC campuses and other location features like proximity to the ocean in our project.

In addition, we decided to do some research into how systemic inequality affects the price of housing in outdated, high-density middle urban neighborhoods. These regions often fall under oppressive zoning laws, predominantly in cities like Los Angeles, which feature checkerboard patterns of wealthy and dilapidated neighborhoods, primarily owing to issues like redlining, gentrification, and poor zoning.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) This structure of inequity may lead to perceived improvements to housing value in our dataset that are not supported by the lived experiences of surrounding residents. 

One notable example of this phenomena is in the neighborhood of South Central Los Angeles. In this region, USC is able to drive economic growth through employment, university expenditures, and private donations. However, the boons brought on by this growth are not shared equitably, as prior residents of the surrounding neighborhoods often suffer from predatory rental contracts, as well as a lack of access to fundamental urban infrastructure like parks, walkable neighborhoods, and shade.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3)

This research ended up influencing what variables we decided to control for–namely, household income, age, and population density. While incomes in an area may rise as a result of the development a University brings, this difference is usually skewed towards new residents who directly benefit from the economic opportunity brought on by the school’s influence, while those who came before are pushed to the side. Ultimately, our decision to control these variables stems from our desire to measure the true effect that Universities have on regional growth, rather than on the difference created by confounding variables like urban displacement and shifts in neighborhood demographics.

## Hypothesis


We expect that census blocks located closer to a UC campus will have higher median house values, especially when the nearest campus is more prestigious. We also expect houses with more bedrooms and those located closer to the ocean to be associated with higher median house values. This is because UC campuses act as major employment and cultural hubs that increase demand for nearby housing, an effect that is expected to be stronger near more prestigious campuses because they are more closely associated with research activity, higher-paying jobs, and sustained investment. Additionally, proximity to the California coastline has historically been associated with higher housing values due to limited supply and high lifestyle demand, while homes with more bedrooms generally have higher prices because they offer more living space.

## Data

The ideal dataset would contain all the variables we mentioned in our research question. However, realistically, it would be best if the dataset would at least contain all the information surrounding the homes, specifically the cost of homes and the features of the homes that we are using as variables, like total bedrooms and proximity to the ocean. Most likely, the dataset would not contain the exact distance to the closest university, so ideally, the dataset would at least include the geographic coordinates, so we can calculate it ourselves. It is not necessary for the dataset to include the prestige of the closest university, since after finding the closest university, we would rely on another source containing the archived rank of these Universities from the 1990 U.S. News College ranking. Ideally, the dataset would contain all census block groups from the 1990 census, which would mean it covers the entire state of California, which is around 20,000 observations. Additionally, the dataset would be collected from the actual data from the 1990 census and be stored in a table, where each row represents a unique census block, with each column being one of the variables, like median house value. 

The first dataset we found is https://www.kaggle.com/datasets/camnugent/california-housing-prices. Luckily for us the dataset contains many of the variables we need from the 1990 census, including c and other ones we can consider such as the total number of rooms. The dataset even includes the output variable we are looking for, median income. The dataset has the longitude and latitude of the houses as well, allowing us to make the calculations for the distance to the closest university. The only issue is the fact that ocean proximity is a categorical variable, which means there might be some difficulty for a multi-variable regression model, however this can be solved easily through one-hot encoding. 

Another dataset we found is https://www.kaggle.com/datasets/fedesoriano/california-housing-prices-data-extra-features. This dataset is almost identical to the last dataset, with a few notable differences. Like the last dataset, it contains the longitude and latitude, the total bedrooms, and the median income, however instead of ocean proximity, it instead uses a quantitative variable called distance to coast, which is the distance to the nearest coast point. It also adds 4 new variables that give the distance to the major cities in California, such as the distance to Los Angeles, or the distance to San Diego.
