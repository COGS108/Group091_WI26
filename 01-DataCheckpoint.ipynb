## Research Question

Does distance from the nearest university in California have the strongest impact on median household income among the others, and how do these factors interact with each other to drive differences in price?

To answer this, we plan to find the Euclidean distance (in miles) between each census block and the nearest university in California using geographic coordinates. ## Background and Prior Work

Housing prices depend on many different factors such as location, income, and housing size. Universities are an important local feature because they attract students, jobs, and other economic activity. In this project, we want to study how strongly housing prices are affected by proximity to a university campus and also a variety of other factors. We also want to examine how well housing prices can be predicted using these factors along with other features such as the number of rooms and proximity to the coastline. 

After researching, we found that UC Riverside looked into a similar question like ours. A group of computer scientists discovered that areas with universities generally have higher home prices and rents than areas without universities.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) However, this study also found that prices near universities tend to rise and fall more quickly. This shows that while universities can influence housing prices, they tend not to be very stable. Our project builds on this work by studying how strongly does proximity to UC campuses relate to housing prices in California compared to other factors.

Previous studies have also looked at how location affects housing prices. Mattera and Franses used methods that group areas based on where they are and how housing prices in those areas move over time. They discovered that models that include location information do a better job at predicting house prices than models that choose to ignore it.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) This also shows how housing prices are influenced not just by house features, but also by where the house is and what is around it. Because of this, we are using distance to university campuses and other location features like proximity to the ocean in our project.

In addition, we decided to do some research into how systemic inequality affects the price of housing in outdated, high-density middle urban neighborhoods. These regions often fall under oppressive zoning laws, predominantly in cities like Los Angeles, which feature checkerboard patterns of wealthy and dilapidated neighborhoods, primarily owing to issues like redlining, gentrification, and poor zoning.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) This structure of inequity may lead to perceived improvements to housing value in our dataset that are not supported by the lived experiences of surrounding residents.

In this region, USC is able to drive economic growth through employment, university expenditures, and private donations. However, the boons brought on by this growth are not shared equitably, as prior residents of the surrounding neighborhoods often suffer from predatory rental contracts, as well as a lack of access to fundamental urban infrastructure like parks, walkable neighborhoods, and shade.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3)

This research ended up influencing what variables we decided to control for–namely, household income, age, and population density. While incomes in an area may rise as a result of the development a University brings, this difference is usually skewed towards new residents who directly benefit from the economic opportunity brought on by the school’s influence, while those who came before are pushed to the side. Ultimately, our decision to control these variables stems from our desire to measure the true effect that Universities have on regional growth, rather than on the difference created by confounding variables like urban displacement and shifts in neighborhood demographics.

This dataset contains a list of all current universities in California founded before 1990, including all defunct universities that closed after that year. Since such a dataset was not readily available online, we ended up creating this dataset manually by scraping University data from wikipedia, and then using mapquest's API to add associated coordinates. After cleaning, the dataset contains 131 institutions and 4 columns, 'Name', 'Founded', 'Latitude', and 'Longitude'. Each row represents one unique institution in California. 

The Latitude and Longitude variables represent the geographic coordinates of each institution measured in decimal degrees. These coordinates will be used to calculate the Euclidean distance between each census block in the housing dataset and the nearest qualifying university. 

There are several important limitations associated with this dataset. First, the dataset is scraped from Wikipedia, which can contain inconsistencies in formatting and information.

Also, geographic coordinates were fetched using an external geocoding API. This introduces potential measurement error. For example, incomplete location information could result in inaccurate latitude and longitude values. Even small coordinate errors can affect calculated distances.

Finally, this dataset includes only institutions listed on Wikipedia and does not necessarily distinguish between research universities, liberal arts colleges, community colleges, or other specialized institutions. These institutions may differ in prestige, economic influence, and other factors. Treating them as on the same level when calculating “distance to nearest university” may oversimplify the mechanisms through which universities affect housing prices.

Despite these limitations, the dataset provides a structured and geographically measurable representation of university locations in California, allowing us to create a proximity-based predictor for analyzing house values in 1990.

```python
## YOUR CODE TO LOAD/CLEAN/TIDY/WRANGLE THE DATA GOES HERE
from bs4 import BeautifulSoup
import requests
import numpy as np
headers = {"User-Agent": "RyanRulkensBot/1.0 (ryanrulkens@gmail.com)"}
uni_data = requests.get("https://en.wikipedia.org/wiki/List_of_colleges_and_universities_in_California", headers=headers).text
soup = BeautifulSoup(uni_data)
schools = soup.find_all("tr")[2:-58]
df = []
for school in schools:
    school_data = school.find_all("td")
    if not school_data:
        continue
    if len(school_data) == 6:
        df.append({"Name": school_data[0].text[:-1], "Founding": school_data[4].text[:-1]})
    else:
        df.append({"Name": school_data[0].text[:-1], "Founding": school_data[3].text[:-1]})
school_data = pd.DataFrame(df)
school_data



```

The above code scrapes data from the List of colleges and universities in California wikipedia page, excluding defunct Universities. These schools are handled in the next code block as they are scraped differently.

```python
defunct_university = []
du = soup.find_all('table')[5].find_all('tr')[1:]
for row in du:
    columns = row.find_all('td')
    closed = columns[3].text.strip().replace('*', '').replace('[2]', '')
    if closed == '':
        continue
    if int(closed) <= 1990:
        continue
    uni = columns[0].text.strip()
    year = columns[2].text.strip().replace('*', '')
    defunct_university.append({"Name": uni, "Founding": year})
defunct_df = pd.DataFrame(defunct_university)
defunct_df.head()
```

The below code concatenates the two dataframes, then clears the Founding column and drops colleges that are too new.

```python
school = pd.concat([school_data, defunct_df])
school_data["Founding"] = school_data["Founding"].apply(lambda x: x.replace("*", "").replace("---", ''))
school_data["Founding"] = school_data["Founding"].apply(lambda x: None if x == "" else int(x))
school_data = school_data[school_data["Founding"] < 1990].reset_index().drop(columns = ["index"])
school_data.head()
```

Here, we use to mapquest api to look up addresses corresponding to the University names in the school_data dataframe. The api then returns the latitude and longitude of these schools, which we add to the school data dataframe. (this cell may take a while to run)

```python
url = "https://www.mapquestapi.com/geocoding/v1/address"
lat_data = []
long_data = []
for university in school_data["Name"]:
    params = {"key": "kH8UIQ6xBOIYz0ABPu1japGd1aUBzFnA", "location": f"{university}"}
    response = requests.get(url, params=params)
    data = response.json()
    latlng = data["results"][0]["locations"][0]["latLng"]
    lat_data.append(latlng['lat'])
    long_data.append(latlng['lng'])
school_data["Latitude"] = pd.Series(lat_data)
school_data["Longitude"] = pd.Series(long_data)
school_data.head()
```
