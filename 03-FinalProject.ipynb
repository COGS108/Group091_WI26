## Abstract

Housing prices are shaped by a complex mix of structural, economic, and geographic factors, but the independent role of university proximity has been relatively underexplored in predictive modeling. This project investigates whether proximity to a California university meaningfully predicts median house value in the 1990 Census, and how that effect compares to other factors like coastal distance, total bedrooms, median housing age, and median income.

To answer this, we scraped a list of 131 California universities founded before 1990 from Wikipedia, geocoded them using the MapQuest API, and used the Haversine formula to calculate the distance from every census block to its nearest campus. We then built multivariate linear regression models using these features, with and without median income as a control variable, to see how much each predictor contributed.

Our models showed that median income was by far the strongest predictor of house value, and that nearest university identity was the second most important feature. Raw distance to university was surprisingly weak and had almost no effect on model performance when removed. Blocks farther from the coast and farther from a university tended to have lower house values, while blocks with more bedrooms, older housing stock, and higher incomes tended to have higher values. The full model achieved an R² of 0.697, compared to 0.498 without income, confirming that much of what looks like a university or coastal effect is actually capturing where wealthier people live. University proximity does relate to housing value, but which university is nearby matters more than how far away it is.

## Research Question

How do distance to the nearest California university, nearest-university identity, total bedrooms, median housing age, and coastal proximity predict median house value in the 1990 California Census dataset, after controlling for median household income? Does university proximity have a statistically meaningful independent effect beyond what income already explains, and does the identity of the nearest university carry additional predictive weight beyond raw distance?

We used total bedrooms measures housing size, median housing age to capture how old the housing stock in a block is, and coastal distance to measure how far a block is from the California coastline. Median household income is included as a control variable because wealthier neighborhoods tend to both attract universities and command higher prices. To compute university distance, we scraped 131 California universities founded before 1990 from Wikipedia, geocoded them using the MapQuest API, and applied the Haversine formula to find each census block's nearest campus.

### Dataset #2

This dataset contains a list of all current universities in California founded before 1990, including all defunct universities that closed after that year. Since such a dataset was not readily available online, we ended up creating this dataset manually by scraping University data from wikipedia, and then using mapquest's API to add associated coordinates. After cleaning, the dataset contains 131 institutions and 4 columns, 'Name', 'Founded', 'Latitude', and 'Longitude'. Each row represents one unique institution in California. 

The Latitude and Longitude variables represent the geographic coordinates of each institution measured in decimal degrees. These coordinates will be used to calculate the Euclidean distance between each census block in the housing dataset and the nearest qualifying university. 

There are several important limitations associated with this dataset. First, the dataset is scraped from Wikipedia, which can contain inconsistencies in formatting and information.

Also, geographic coordinates were fetched using an external geocoding API. This introduces potential measurement error. For example, incomplete location information could result in inaccurate latitude and longitude values. Even small coordinate errors can affect calculated distances.

Finally, this dataset includes only institutions listed on Wikipedia and doesn't necessarily distinguish between research universities, liberal arts colleges, community colleges, or other specialized institutions. These institutions may differ in prestige, economic influence, and other factors. Treating them as on the same level when calculating “distance to nearest university” may oversimplify the mechanisms through which universities affect housing prices.

Despite these limitations, the dataset provides a structured and geographically measurable representation of university locations in California, allowing us to create a proximity-based predictor for analyzing house values in 1990.

```python
## YOUR CODE TO LOAD/CLEAN/TIDY/WRANGLE THE DATA GOES HERE
from bs4 import BeautifulSoup
import requests
import numpy as np
headers = {"User-Agent": "RyanRulkensBot/1.0 (ryanrulkens@gmail.com)"}
uni_data = requests.get("https://en.wikipedia.org/wiki/List_of_colleges_and_universities_in_California", headers=headers).text
soup = BeautifulSoup(uni_data)
schools = soup.find_all("tr")[2:-58]
df = []
for school in schools:
    school_data = school.find_all("td")
    if not school_data:
        continue
    if len(school_data) == 6:
        df.append({"Name": school_data[0].text[:-1], "Founding": school_data[4].text[:-1]})
    else:
        df.append({"Name": school_data[0].text[:-1], "Founding": school_data[3].text[:-1]})
school_data = pd.DataFrame(df)
school_data
```

The above code scrapes data from the List of colleges and universities in California wikipedia page, excluding defunct Universities. These schools are handled in the next code block as they are scraped differently.

```python
defunct_university = []
du = soup.find_all('table')[5].find_all('tr')[1:]
for row in du:
    columns = row.find_all('td')
    closed = columns[3].text.strip().replace('*', '').replace('[2]', '')
    if closed == '':
        continue
    if int(closed) <= 1990:
        continue
    uni = columns[0].text.strip()
    year = columns[2].text.strip().replace('*', '')
    defunct_university.append({"Name": uni, "Founding": year})
defunct_df = pd.DataFrame(defunct_university)
defunct_df.head()
```

The below code concatenates the two dataframes, then clears the Founding column and drops colleges that are too new.

```python
school_data = pd.concat([school_data, defunct_df])
school_data["Founding"] = school_data["Founding"].apply(lambda x: x.replace("*", "").replace("---", ''))
school_data["Founding"] = school_data["Founding"].apply(lambda x: None if x == "" else int(x))
school_data = school_data[school_data["Founding"] < 1990].reset_index().drop(columns = ["index"])
school_data['Founding'] = school_data['Founding'].astype(int)
school_data.head()
```

Here, we use to mapquest api to look up addresses corresponding to the University names in the school_data dataframe. The api then returns the latitude and longitude of these schools, which we add to the school data dataframe. (this cell may take a while to run)

```python
url = "https://www.mapquestapi.com/geocoding/v1/address"
lat_data = []
long_data = []
for university in school_data["Name"]:
    params = {"key": "kH8UIQ6xBOIYz0ABPu1japGd1aUBzFnA",
              "location": f"{university}, California", 
              "boundingBox": "42.01,-124.48,32.53,-114.13",
              "maxResults": 1}
    response = requests.get(url, params=params)
    data = response.json()
    latlng = data["results"][0]["locations"][0]["latLng"]
    lat_data.append(latlng['lat'])
    long_data.append(latlng['lng'])
school_data["Latitude"] = pd.Series(lat_data)
school_data["Longitude"] = pd.Series(long_data)
school_data.head()
```

### Section 2: Spatial Analysis of University Proximity

In this section, we combined our two primary datasets by calculating the geographic distance between each census block and the nearest institution of higher education. First we need to remove defunct universities, or ones that did not exist in 1990, and manually correct the latitude and longitudes of some universities that were outliers. 

```python
# removed defunct universities from school
school_data = school_data[~school_data['Name'].isin(defunct_df['Name'])]

# correct latitude and longitudes of some universities
corrections = {
    "California State University, Stanislaus": (37.5256, -120.8548),
    "Fuller Theological Seminary": (34.14853613, -118.1404425),
    "Chapman University": (37.52331378, -120.8582655),
    "Harvey Mudd College": (34.1063171, -117.7084285),
    "Pacific School of Religion": (37.87684384, -122.263019),
    "Hult International Business School": (37.80276594, -122.4032164),
    "Pacific Lutheran Theological Seminary": (37.87008372, -122.2702566),
    "Institute of Buddhist Studies": (37.86695056, -122.2659889),
    "NewSchool of Architecture and Design": (32.71836013, -117.1587197),
    "Charles R. Drew University of Medicine and Science": (33.92563, -118.24259)
}


def apply_corrections(df, corrections_dict):
    for school, coords in corrections_dict.items():
        df.loc[df['Name'] == school, ['Latitude', 'Longitude']] = coords[0], coords[1]
    return df

school_data = apply_corrections(school_data, corrections)

outliers = school_data[(school_data["Longitude"] > -114) | (school_data["Longitude"] < -125)]
print(f"Remaining outliers: {len(outliers)}")
```

In this stage of the EDA, we transformed our raw geographic coordinates into a functional predictor: Distance to University. By applying the Haversine formula, which accounts for the Earth's curvature, we accurately mapped each census block to its closest 1990-era university. This allows us to begin testing the effect of distance to university on housing prices.

```python
# In this cell calculated university with shortest distance and added a new column to houses that has the closest university to it and the distance in meters to that university; 
# We also made different plots

# this func calculates the distance between two points given lat and long for both on a sphere (Earth)
def haversine(lat1, lon1, lat2, lon2):  
    R = 6371  # Earth radius in km
    lat1 = np.radians(lat1)
    lon1 = np.radians(lon1)
    lat2 = np.radians(lat2)
    lon2 = np.radians(lon2)
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = np.sin(dlat/2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon/2)**2
    c = 2 * np.arcsin(np.sqrt(a))
    return R * c    #distance returned in kilometers

school_lats = school_data["Latitude"].values
school_lons = school_data["Longitude"].values
school_names = school_data["Name"].values
closest_university = []
closest_distance_meters = []
for _, house in houses.iterrows():
    distances = haversine(
        house["Latitude"],
        house["Longitude"],
        school_lats,
        school_lons
    )
    min_index = np.argmin(distances)
    closest_university.append(school_names[min_index])
    closest_distance_meters.append(distances[min_index] * 1000)
houses["Closest_University"] = closest_university
houses["Distance_to_University_m"] = np.round(closest_distance_meters, 0)




import matplotlib.cm as cm


# Plot 1: Top 10 Universities with Most Nearby Houses
uni_counts = houses['Closest_University'].value_counts().head(10)
fig1, ax1 = plt.subplots(figsize=(10,6))
colors = cm.viridis(np.linspace(0,1,len(uni_counts)))
ax1.barh(uni_counts.index, uni_counts.values, color=colors, edgecolor='black')
ax1.set_xlabel("Number of Houses")
ax1.set_title("Top 10 Universities with Most Nearby Houses", fontsize=14)
ax1.invert_yaxis()
ax1.grid(axis='x', linestyle='--', alpha=0.7)
fig1.tight_layout()
```

This bar chart identifies the primary geographic hubs of our 1990 dataset, showing that CSU Northridge and CSU Fresno serve as the nearest institutions for the largest clusters of census blocks. ## Analysis: Modeling Housing Prices Across California

```python
#Imports for ML models
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import FunctionTransformer
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.pipeline import Pipeline
```

Before training any models, we decided to check whether any of the quantitative features we are using (total bedrooms, distance to coast, distance to university, total households, median income, and median age) are strongly correlated with each other. We do this to find if any of the variables show high multicollinearity, which would increase variance and lower model interpretability.

```python
# calculating correlations between each variable
corr = houses[["Tot_Bedrooms", "Distance_to_coast", "Distance_to_University_m", "Households", "Median_Income", "Median_Age"]].corr()

# Plotting the correlations in a heatmap visualization
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Heatmap")
plt.show()
```

From this, we found that total households and total bedrooms and strongly correlated with each other, and so though we initially added number of households to our model as a control, we decide to drop that column for all our models. When training our first model, we decide to one hot encode the closest university column, dropping one column to avoid linear dependence. We also decided to log transform the two distance columns, as these two showed a non-linear relationship with median house value in our EDA section. The remaining columns were then standardized to ensure that features with large values, like total bedrooms, would not be disproportionately valued in our model. This model ended up producing an R-squared value of 0.697, and a RMSE of 53611.

```python
to_train = houses.copy() 
X = to_train[["Tot_Bedrooms", "Distance_to_coast", "Distance_to_University_m", "Closest_University", "Median_Income", "Median_Age"]] 
y = to_train["Median_House_Value"]

# splitting the data into train and test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state = 42) 


# one hot encoding closest university column and transforming distance columns using np.log1p
preproc = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(drop="first"), ["Closest_University"]),
        ('log', FunctionTransformer(np.log1p), ["Distance_to_coast", "Distance_to_University_m"]),
    ],
    remainder=StandardScaler())


# creating pipeline with LinearRegression as the model
pl = Pipeline([('preprocess', preproc), ('lin-reg', LinearRegression())])


pl.fit(X_train, y_train) # fitting or training the model
print("Score for best model: " + str(pl.score(X_test, y_test))) # score of our model using R2; closer to 1 means better
print("RMSE for best model: " + str(np.sqrt(mean_squared_error(y_test, pl.predict(X_test)))))

# plotting the actual y-values and then plotting predicted y-values to compare
# x-values are not real predictors but just the index-number/datapoint of a row in houses
to_train["Median_House_Value_Predicted"] = pl.predict(X)
ax = to_train[:500].plot(kind="line", y="Median_House_Value")
to_train[:500].plot(kind="line", y="Median_House_Value_Predicted", ax=ax)
ax.set_title("Actual vs Predicted Median House Value")
ax.set_xlabel("Index")
ax.set_ylabel("Median House Value")
```

It seems that our model predicts the true value of houses quite well! 

#### Analyzing Which Variable has the Highest Predictive Power

In the next step, we wanted to see which features were the strongest predictors for the variation in median house value. To do this, we created a new model using only the quantative variables (bedrooms, median income, median age, and the distances), then applied the same transformations to these columns before fitting them to the data. After training the model, we then pulled out the coeffients associated with each feature, and thanks to standardizing the data beforehand, were able to see that while the columns distance to coast and median income were strong predictors of housing value, distance to university actually proved to be a pretty weak variable. Still, the fact that the coefficient is negative does prove that the further you get from a university, the lower housing prices tend to be.

```python
to_train = houses.copy() 
X = to_train[["Tot_Bedrooms", "Distance_to_coast", "Distance_to_University_m", "Median_Income", "Median_Age"]] #only used quantitative variables
y = to_train["Median_House_Value"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state = 42)
preproc = ColumnTransformer(
    transformers=[
        ('log_std', Pipeline([
            ('log', FunctionTransformer(np.log1p, feature_names_out="one-to-one")), # log transforming distance variables
            ('scale', StandardScaler())                                             # standarding the distance variables
        ]), ["Distance_to_coast", "Distance_to_University_m"])
    ],
    remainder=StandardScaler())                                                     # only just standardizing the other variables
pl = Pipeline([('preprocess', preproc), ('lin-reg', LinearRegression())])
pl.fit(X_train, y_train)


# creating a table with each row being a feature and a column being the optimal coefficient
coefs = pl.named_steps["lin-reg"].coef_
features = pl.named_steps["preprocess"].get_feature_names_out()
coef_df = pd.DataFrame(coefs, index=features, columns=["Coefficient"])
display(coef_df)


# creating a bar chart for model coefficients
coef_df_reset = coef_df.reset_index()
coef_df_reset.columns = ["Feature", "Coefficient"]
plt.figure(figsize=(10, 6))
sns.barplot(data=coef_df_reset, x="Coefficient", y="Feature", palette="coolwarm", hue="Feature", legend=False)
plt.title("Model Coefficients")
plt.axvline(0, color='black', linewidth=1)
plt.show()
```

After assessing importance of features for the quantitative variables, we then decided to add on the categorical variable of closest university name as well. In the cell below, we tried dropping one column at a time from the original model, then applying the same transformations to the remaining columns and computing a R squared value for each model. From this step, we are able to see that median income and closest university are the most important features for upholding model performance, while distance to university once again proves to be the least useful feature as removing it as a predictor basically has no effect on performance.

```python
cols = ["Tot_Bedrooms", "Distance_to_coast", "Distance_to_University_m", "Median_Age", "Closest_University", "Median_Income"] #added back categorical variable Closest University
r2_scores = []

# calculating the R Squared Scores for each model by dropping a different variable each time
for col in cols:
    X_temp = to_train[cols].drop(columns=col)   
    y = to_train["Median_House_Value"] 
    X_train, X_test, y_train, y_test = train_test_split(X_temp, y, test_size=0.25, random_state=42)  
    preproc = ColumnTransformer(
        transformers=[
            ('cat', OneHotEncoder(drop="first"), ["Closest_University"] if "Closest_University" in X_temp.columns else []),
            ('sqrt', FunctionTransformer(np.sqrt),
             [c for c in ["Distance_to_coast", "Distance_to_University_m"] if c in X_temp.columns])],
        remainder=StandardScaler())   
    pl = Pipeline([('preprocess', preproc), ('lin-reg', LinearRegression())])    
    pl.fit(X_train, y_train)
    r2_scores.append(pl.score(X_test, y_test))

# plotting the scores of each model trained on dropping that certain variable
results = pd.DataFrame({"Dropped_Feature": cols, "R2_Score": r2_scores}).sort_values(by="R2_Score", ascending=False)
results.plot(kind="bar", x="Dropped_Feature", xlabel = "Dropped Feature", y="R2_Score", ylabel="R2 Score", legend=False, title="R Squared of Model with Each Feature Dropped")
```

While we initially included median income to our model as a control, it turns out that this feature is actually the strongest predictor. This isn't necessarily a bad thing though, is it shows that it was right to include income as a control to see the true (albeit weaker) signal in the other variables. Training a model without controlling for median income, we can see from the R squared and RMSE that the model's predictive power suffers greatly, while the graph shows that the model is a lot less reactive to variation in the data.

```python
to_train = houses.copy() 
X = to_train[["Tot_Bedrooms", "Distance_to_coast", "Distance_to_University_m", "Median_Age", "Closest_University"]]  # dropped Median_Income
y = to_train["Median_House_Value"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state = 42)
preproc = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(drop="first"), ["Closest_University"]),
        ('sqrt', FunctionTransformer(np.log1p), ["Distance_to_coast", "Distance_to_University_m"]),
    ],
    remainder=StandardScaler())
pl = Pipeline([('preprocess', preproc), ('lin-reg', LinearRegression())])
pl.fit(X_train, y_train)
print("Score for model without median income: " + str(pl.score(X_test, y_test)))
print("RMSE for best model: " + str(np.sqrt(mean_squared_error(y_test, pl.predict(X_test)))))


# plotting the actual y-values and then plotting predicted y-values to compare for this model without Median_Income
# x-values are not real predictors but just the index-number/datapoint of a row in houses
to_train["Median_House_Value_Predicted"] = pl.predict(X)
ax = to_train[:500].plot(kind="line", y="Median_House_Value")
to_train[:500].plot(kind="line", y="Median_House_Value_Predicted", ax=ax)
ax.set_title("Actual vs Predicted Median House Value")
ax.set_xlabel("Index")
ax.set_ylabel("Median House Value")
```

# Improving Model
We now try to make our model accuracy better. In order to do that, we first decided to transform our y-values by log transforming them. Then, we also decided to try out different models, hyperparameters, transformations, and k-fold cross-validation technique. By using the k-fold cross-validation technique and tuning hyperparameters, it will help our model generalize and perform better with new/testing data. To do all this, we used GridSearchCV that tests out all possible combinations and finds the best model, hyperparameters, and transformations. 

```python
# y-values (log transform)
y = np.log1p(houses["Median_House_Value"])


# Variables/Features
X = houses[["Distance_to_University_m", "Tot_Bedrooms", "Distance_to_coast", "Closest_University", "Median_Income", "Median_Age"]]

# splitting data into test and train
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)


# grouping the features
distance_features = ["Distance_to_University_m", "Distance_to_coast"]
numeric_features = ["Tot_Bedrooms", "Median_Income", "Median_Age"]
categorical_features = ["Closest_University"]


# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ("distance", "passthrough", distance_features),
        ("num", "passthrough", numeric_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features),
    ]
)


# pipline
pipeline = Pipeline(
    [
        ("preprocess", preprocessor),
        ("poly", PolynomialFeatures(include_bias=False)),
        ("scaler", StandardScaler(with_mean=False)),
        ("model", LinearRegression()),
    ]
)



# transformer for distance
log_transform = FunctionTransformer(np.log1p)


# grid search parameters
param_grid = [

    # Trying all combinations of Linear Regression model
    {
        "preprocess__distance": [
            "passthrough",
            Pipeline([("log", log_transform)])
        ],
        "poly__degree": [1, 2],
        "poly__interaction_only": [False, True],
        "scaler": [StandardScaler(with_mean=False), "passthrough"],
        "model": [LinearRegression()]
    },

    # Trying all combinations of Ridge model
    {
        "preprocess__distance": [
            "passthrough",
            Pipeline([("log", log_transform)])
        ],
        "poly__degree": [1, 2],
        "poly__interaction_only": [False, True],
        "scaler": [StandardScaler(with_mean=False), "passthrough"],
        "model": [Ridge()],
        "model__alpha": [0.1, 1, 10]
    },
]


# Grid Search
grid = GridSearchCV(pipeline, param_grid, cv=5, scoring="r2", n_jobs=-1, verbose=2,)
grid.fit(X_train, y_train)

# best results
print("\nBest Parameters:")
print(grid.best_params_)

print("\nBest CV Score:")
print(grid.best_score_)

best_model = grid.best_estimator_
test_r2 = best_model.score(X_test, y_test)

print("\nTest R2:", test_r2)
```

The GridSearchCV helped us find that best model is Ridge with specific transformations and hyperparameters. By also doing k-fold cross-validation, it also helped us increase our test accuracy by bumping it up to around 0.784, which means that our model generalizes well.


```python
# plotting predictions vs actual again for final model
model2_table = houses.copy()
model2_table['predicts'] = np.expm1(best_model.predict(X))
ax = model2_table[:200].plot(kind='line', y='Median_House_Value')
model2_table[:200].plot(kind='line', y='predicts', ax=ax)
ax.set_title("Actual vs Predicted Median House Value")
ax.set_xlabel("Index")
ax.set_ylabel("Median House Value")
```

# Discussion and Conclusion

Our full model, which included total bedrooms, median housing age, distance to coast, distance to university, closest university, and median income, achieved an R² of 0.697 and an RMSE of 53,611 on the test set. Looking at the model coefficients, distance to coast and distance to university both had negative values, meaning blocks farther from the coast and farther from a university tended to have lower predicted house values. Total bedrooms, median housing age, and median income all had positive coefficients, consistent with what we saw in the EDA and with our original hypotheses.

The drop-one analysis showed that median income was the strongest predictor by far, dropping R² from 0.697 to 0.498 when removed. Closest university was the second most important feature, dropping R² to around 0.57, while dropping raw distance to university had almost no effect on performance. Looking at specific schools, Lincoln University, Stanford, and UC Berkeley neighborhoods had the highest average house values, while CSU Northridge and CSU Fresno had the most nearby census blocks but much more moderate prices.
The bigger picture here is that university proximity does relate to housing value, but mostly as a marker of which neighborhood you're in rather than as a direct driver of prices. High-value universities like Stanford and UC Berkeley are located in already-wealthy Bay Area neighborhoods, so it's hard to separate the university effect from the broader socioeconomic conditions of those areas. This is likely why income dominated the model and why coastal distance was also a strong signal, since all three tend to cluster together geographically.

One result that was surprising was how little raw distance to university mattered once we knew which university was closest. We expected distance to be one of our main predictors, but removing it barely changed model performance at all. This is probably because most census blocks are already quite close to some university, since the distribution was heavily right-skewed with over 10,000 blocks within 10,000 meters of a campus, so there's not much variation in distance to exploit. What varied meaningfully was which school was nearby.
These results extend the UC Riverside finding that university-adjacent areas have higher prices by showing the effect is highly institution-dependent. The type of university and the neighborhood it sits in matters far more than just proximity.

There are a few limitations worth noting. Our dataset is a single snapshot from 1990, so we can't say whether these patterns hold over time. The university variable has many categories and some schools appear near very few census blocks, so those coefficients might not be reliable. Finally, since all our predictors are observational and spatially correlated, we can't make causal claims. Some next steps might be to bring in additional neighborhood-level features like school district quality or transit access to reduce how much our university coefficient is picking up other things. Comparing results across different years would also help determine whether the university proximity effect has grown stronger over time as California's economy has shifted.
